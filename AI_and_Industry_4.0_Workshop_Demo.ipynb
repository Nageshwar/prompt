## 0. Environment Setup & Local AI Health Check
Verify network dependencies and check if the local Ollama instance is available.

```python
%pip install requests beautifulsoup4
```

```python
import requests
from bs4 import BeautifulSoup

OLLAMA_BASE_URL = "http://localhost:11434"
DEFAULT_MODEL = "qwen2.5:3b"

def query_ollama(prompt, system_prompt="", model=DEFAULT_MODEL):
    """Query the local Ollama instance via its REST API."""
    url = f"{OLLAMA_BASE_URL}/api/generate"
    payload = {
        "model": model,
        "prompt": prompt,
        "system": system_prompt,
        "stream": False
    }
    try:
        response = requests.post(url, json=payload, timeout=60)
        if response.status_code == 200:
            return response.json().get("response", "")
        return f"Ollama Error {response.status_code}: {response.text}"
    except requests.exceptions.RequestException:
        return ("[Offline Demo Mode] Ollama request failed or timed out.\n"
                "Ensure Ollama is started with `ollama run qwen2.5:3b` in your terminal.")

# Check server status
print("Checking local Ollama connection...")
try:
    res = requests.get(f"{OLLAMA_BASE_URL}/api/tags", timeout=2)
    models = [m['name'] for m in res.json().get('models', [])]
    print("✅ Ollama is running! Available local models:", models)
except Exception:
    print("ℹ️ Ollama is currently offline. Prompts below can still be reviewed and will use local fallback demos.")
```

```python
building_blocks_prompt = """
[Persona]: Act as head of the College Robotics Club.
[Objective]: Write a WhatsApp announcement recruiting 1st and 2nd year students for our Hackathon.
[Example]:
Notice: "Meeting at 5pm room 204."
-> "🤖 Builders Wanted! Kickstart your coding journey at 5 PM in Room 204. Free pizza!"
[Format]:
- Catchy headline with 1 emoji
- 2 bullet points on benefits
- Meeting date, time & room
- Call to Action button link
[Tone]: Energetic, welcoming to beginners, zero jargon.
"""

print("=== PROMPT INPUT ===")
print(building_blocks_prompt.strip())
print("\n=== AI OUTPUT ===")
out = query_ollama(building_blocks_prompt)
if "[Offline Demo Mode]" in out:
    print("""🚀 Build Robots, Win Prizes: Robotics Club 2026!

No prior coding experience needed—if you love tinkering, you belong here!

• Learn Arduino, 3D printing & AI from senior mentors.
• Build real hardware projects for inter-college hackathons.

📍 Where: Makerspace Lab (Room 304)
🗓️ When: Thursday, Oct 8th @ 4:30 PM

👉 Register your spot: bit.ly/ssm-robotics-join""")
else:
    print(out)
```

## 2. Foundational Techniques: CoT & ReAct Agent Loop
Demonstrating step-by-step reasoning and autonomous tool-using cycles.

```python
# Chain-of-Thought (CoT) Calculation
cot_prompt = """
Calculate roommate Wi-Fi bill split step-by-step.
Total bill is $60 for the month of January (30 days total).
- Roommate Alex stayed all 30 days.
- Roommate Sam went home for winter break and stayed only 10 days.
- Roommate Chris stayed 20 days.

Think step-by-step to calculate how much each person owes fairly based on person-days.
"""
print("=== Chain-of-Thought Prompt ===")
print(cot_prompt.strip())
print("\n=== Reasoning Output ===")
cot_out = query_ollama(cot_prompt)
if "[Offline Demo Mode]" in cot_out:
    print("""Step 1: Calculate total person-days:
• Alex: 30 days | Sam: 10 days | Chris: 20 days
Total = 30 + 10 + 20 = 60 person-days.

Step 2: Calculate daily rate per person-day:
• $60 total bill / 60 person-days = $1.00 per person per day.

Step 3: Multiply each person's days by daily rate:
• Alex owes: 30 * $1.00 = $30.00
• Sam owes: 10 * $1.00 = $10.00
• Chris owes: 20 * $1.00 = $20.00

Final Balance: Alex: $30 | Sam: $10 | Chris: $20 (Total = $60.00).""")
else:
    print(cot_out)
```

```python
# ReAct Agent Loop: Simulated Tool Calling
def catalog_search(title):
    database = {
        "clean code": {"barcode": "98231", "shelf": "4B", "status": "Available"},
        "deep learning": {"barcode": "77123", "shelf": "1A", "status": "Checked Out"}
    }
    return database.get(title.lower(), {"status": "Not Found"})

def place_reservation(barcode, student_id):
    return f"Reservation active for ID {student_id} on Barcode #{barcode}. Valid for 24 hours."

# ReAct execution simulation
target_book = "Clean Code"
student = "SSM-2026-104"

print("=== ReAct Execution Trace ===")
print(f"Thought: Searching catalog database for '{target_book}'...")
info = catalog_search(target_book)
print(f"Action: catalog_search('{target_book}') -> Observation: {info}")

if info.get("status") == "Available":
    print(f"Thought: Book is on shelf {info['shelf']}. Placing active reservation hold...")
    confirmation = place_reservation(info['barcode'], student)
    print(f"Action: place_reservation({info['barcode']}, {student}) -> Observation: {confirmation}")
    print(f"\nFinal Output: Success! '{target_book}' is reserved for student {student} at Main Desk (Shelf {info['shelf']}).")
```

## 4. Live Campus Automation Agents (`ssmahavidyalaya.edu.in`)
Running the 4 agents featured in the workshop.

```python
# Agent 1: Smart Notice Board Radar
def fetch_ssm_notices():
    """Scrapes official notices or falls back gracefully."""
    url = "https://ssmahavidyalaya.edu.in/blog/"
    try:
        resp = requests.get(url, timeout=5, headers={"User-Agent": "Mozilla/5.0"})
        soup = BeautifulSoup(resp.text, 'html.parser')
        titles = [h.get_text().strip() for h in soup.find_all(['h2', 'h3'], limit=3)]
        if titles:
            return titles
    except Exception:
        pass
    return ["Swami Vivekananda Merit-cum-Means (SVMCM) Scholarship Renewal verification at office before Oct 25, 2026."]

latest = fetch_ssm_notices()[0]
print("Latest Notice Detected:", latest)

radar_prompt = f"""
Act as an academic notice triage bot for Sukumar Sengupta Mahavidyalaya.
Notice Text: "{latest}"
Extract strict JSON with fields: 'title', 'category', 'summary', 'deadline'.
"""
print("\n--- AI Triage JSON ---")
print(query_ollama(radar_prompt))
```

```python
# Agent 2: Holiday & Campus Closure Tracker
holiday_text = """
SSM Circular: College will observe Durga Puja & Diwali Vacation from 18th October 2026 
to 3rd November 2026. Normal administrative and academic activities resume on 4th November 2026.
"""

cal_prompt = f"""
Act as a calendar event extractor.
Circular: "{holiday_text}"
Output JSON schema: {{"event": "...", "start": "YYYY-MM-DD", "end": "YYYY-MM-DD", "reopen": "YYYY-MM-DD"}}
"""
print("--- Calendar Node Extraction ---")
print(query_ollama(cal_prompt))
print("\n✅ Event scheduled in Google Calendar Node with reminder set for Nov 3rd.")
```
